```python
from stabilizer_circuit import RepititionCode
import matplotlib.pyplot as plt
```

```python
results = {}
bins = 30
for i in range(bins):
    error_prob = i/bins
    code = RepititionCode(3)
    error_rate = code.run(error_prob=error_prob, error_weight=3, shots=1000, random_initial_state=True)
    results[error_prob] = error_rate

plt.figure()
plt.plot(results.keys(), results.values())
plt.show()
```

```python
results_d = {}
bins_d = 1000
for i in range(bins_d):
    p = i/bins_d
    res = 3*((1 - p)*p*p) + p**3  # 3p^2 - 2p^3 
    results_d[p] = res

plt.figure()
plt.plot(results_d.keys(), results_d.values())
plt.show()
```
